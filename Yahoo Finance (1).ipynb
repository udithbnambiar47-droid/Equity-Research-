### Basic yfinance Option Chain Commands

| Command                                                   | Purpose                                                    |
| --------------------------------------------------------- | ---------------------------------------------------------- |
| `import yfinance as yf`                                   | Import the `yfinance` library.                             |
| `stock = yf.Ticker("AAPL")`                               | Create a ticker object for a stock.                        |
| `stock.options`                                           | Display all available option expiration dates.             |
| `stock.option_chain()`                                    | Retrieve the option chain for the nearest expiration date. |
| `stock.option_chain("2026-08-21")`                        | Retrieve the option chain for a specific expiration date.  |
| `option.calls`                                            | Display the call options DataFrame.                        |
| `option.puts`                                             | Display the put options DataFrame.                         |
| `calls.head()`                                            | Show the first 5 call option contracts.                    |
| `puts.head()`                                             | Show the first 5 put option contracts.                     |
| `calls.columns`                                           | List all available columns in the call options DataFrame.  |
| `calls[['strike','lastPrice']]`                           | Display strike prices and last traded prices.              |
| `calls.sort_values("volume", ascending=False)`            | Sort call options by trading volume.                       |
| `calls.sort_values("openInterest", ascending=False)`      | Sort call options by open interest.                        |
| `calls.sort_values("impliedVolatility", ascending=False)` | Sort call options by implied volatility.                   |
| `calls[calls["inTheMoney"] == True]`                      | Display in-the-money (ITM) call options.                   |
| `calls[calls["inTheMoney"] == False]`                     | Display out-of-the-money (OTM) call options.               |
| `puts[calls["inTheMoney"] == True]`                       | Display in-the-money put options.                          |
| `puts[calls["inTheMoney"] == False]`                      | Display out-of-the-money put options.                      |
| `stock.fast_info["lastPrice"]`                            | Get the current stock price.                               |
| `stock.fast_info["bid"]`                                  | Get the current bid price (if available).                  |
| `stock.fast_info["ask"]`                                  | Get the current ask price (if available).                  |
| `calls.to_csv("calls.csv")`                               | Save call options to a CSV file.                           |
| `puts.to_csv("puts.csv")`                                 | Save put options to a CSV file.                            |
| `calls.to_excel("calls.xlsx")`                            | Save call options to an Excel file.                        |
| `puts.to_excel("puts.xlsx")`                              | Save put options to an Excel file.                         |
| `for expiry in stock.options:`                            | Loop through all available expiration dates.               |
| `stock.option_chain(expiry)`                              | Get the option chain for each expiration date in a loop.   |


### Common Option Chain Columns

| Column              | Purpose                                                                              |
| ------------------- | ------------------------------------------------------------------------------------ |
| `contractSymbol`    | Unique identifier of the option contract.                                            |
| `strike`            | Strike price of the option.                                                          |
| `lastPrice`         | Last traded option price.                                                            |
| `bid`               | Highest current bid price.                                                           |
| `ask`               | Lowest current ask price.                                                            |
| `change`            | Price change since the previous trading session.                                     |
| `percentChange`     | Percentage price change.                                                             |
| `volume`            | Number of contracts traded today.                                                    |
| `openInterest`      | Number of outstanding contracts.                                                     |
| `impliedVolatility` | Market-implied volatility of the option.                                             |
| `inTheMoney`        | Indicates whether the option is currently in the money (`True`/`False`).             |
| `lastTradeDate`     | Date and time of the last trade.                                                     |
| `contractSize`      | Contract size (typically `REGULAR`, representing 100 shares in U.S. equity options). |
| `currency`          | Currency in which the option is traded.                                              |


**Note: Option chain data through yfinance is generally available for U.S. stocks (e.g., AAPL, MSFT, TSLA). Yahoo Finance typically does not provide NSE option chain data (e.g., TCS.NS, RELIANCE.NS) via yfinance.**

# Convert Scientic Notation Number to Normal Number

**yfinance returns many large financial values (such as Market Cap, Revenue, Cash Flow, Total Assets, and Debt) as Python float values. Python and Pandas often display large numbers in scientific notation to save space.**

 Market Cap = 3.2456e+12

**This means: 3,245,600,000,000**

# Method 1: Convert to an Integer (Best for Finance) 

```python
market_cap = 3.2456e+12
```

```python
print(int(market_cap))
```

# Method 2: Format with Commas 

```python
print(f"{market_cap:,.0f}")
```

# Method 3: Format Without Commas

```python
print(f"{market_cap:.0f}")
```
