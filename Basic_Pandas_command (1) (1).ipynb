# The most important NumPy commands for Data Analysts and Data Scientists.

| Command                      | Purpose                                                                                          |
| ---------------------------- | ------------------------------------------------------------------------------------------------ |
| `df.head()`                  | Display the first 5 rows of the DataFrame.                                                       |
| `df.tail()`                  | Display the last 5 rows of the DataFrame.                                                        |
| `df.shape`                   | Return the number of rows and columns as `(rows, columns)`.                                      |
| `df.info()`                  | Show dataset information, including column names, data types, non-null counts, and memory usage. |
| `df.describe()`              | Generate summary statistics for numeric columns.                                                 |
| `df.columns`                 | Display all column names.                                                                        |
| `df.dtypes`                  | Show the data type of each column.                                                               |
| `df.index`                   | Display the row index labels.                                                                    |
| `df.isnull()`                | Check for missing (`NaN`) values.                                                                |
| `df.isnull().sum()`          | Count missing values in each column.                                                             |
| `df.notnull()`               | Check for non-missing values.                                                                    |
| `df.duplicated()`            | Identify duplicate rows.                                                                         |
| `df.duplicated().sum()`      | Count the number of duplicate rows.                                                              |
| `df.dropna()`                | Remove rows containing missing values.                                                           |
| `df.fillna(value)`           | Replace missing values with a specified value.                                                   |
| `df.drop_duplicates()`       | Remove duplicate rows.                                                                           |
| `df.sort_values()`           | Sort rows by one or more columns.                                                                |
| `df.sort_index()`            | Sort the DataFrame by index.                                                                     |
| `df["Column"]`               | Select a single column.                                                                          |
| `df[["Col1", "Col2"]]`       | Select multiple columns.                                                                         |
| `df.loc[]`                   | Select rows and columns by labels.                                                               |
| `df.iloc[]`                  | Select rows and columns by integer positions.                                                    |
| `df.query()`                 | Filter rows using a query expression.                                                            |
| `df.sample()`                | Return random rows from the DataFrame.                                                           |
| `df.groupby()`               | Group data for aggregation and analysis.                                                         |
| `df.agg()`                   | Apply one or more aggregate functions.                                                           |
| `df.merge()`                 | Join two DataFrames based on common columns.                                                     |
| `pd.concat()`                | Combine multiple DataFrames vertically or horizontally.                                          |
| `df.join()`                  | Join DataFrames using their indexes.                                                             |
| `df.value_counts()`          | Count the frequency of unique values.                                                            |
| `df.nunique()`               | Count the number of unique values.                                                               |
| `df.unique()`                | Return unique values from a column.                                                              |
| `df.sum()`                   | Calculate the sum of values.                                                                     |
| `df.mean()`                  | Calculate the average value.                                                                     |
| `df.median()`                | Calculate the median value.                                                                      |
| `df.min()`                   | Return the minimum value.                                                                        |
| `df.max()`                   | Return the maximum value.                                                                        |
| `df.std()`                   | Calculate the standard deviation.                                                                |
| `df.count()`                 | Count non-missing values.                                                                        |
| `df.apply()`                 | Apply a function to rows or columns.                                                             |
| `df.astype()`                | Convert a column to a different data type.                                                       |
| `pd.to_numeric()`            | Convert values to numeric type.                                                                  |
| `pd.to_datetime()`           | Convert values to date/time format.                                                              |
| `df.rename()`                | Rename columns or index labels.                                                                  |
| `df.drop()`                  | Remove rows or columns.                                                                          |
| `df.set_index()`             | Set a column as the index.                                                                       |
| `df.reset_index()`           | Reset the index to the default integer index.                                                    |
| `df.memory_usage()`          | Show memory usage of each column.                                                                |
| `df.memory_usage(deep=True)` | Show accurate memory usage, including string objects.                                            |
| `df.corr()`                  | Calculate the correlation between numeric columns.                                               |
| `df.to_csv()`                | Save the DataFrame to a CSV file.                                                                |
| `df.to_excel()`              | Save the DataFrame to an Excel file.                                                             |
| `pd.read_csv()`              | Read data from a CSV file.                                                                       |
| `pd.read_excel()`            | Read data from an Excel file.                                                                    |
